### 1) Problem Statement

* This project aims to develop an Artificial Intelligence system capable of recognizing human emotions from facial images and video streams.

#### The system identifies emotional states such as:
* Angry
* Disgust
* Fear
* Happy
* Sad
* Surprise
* Neutral
#### The model can be used in applications such as:
* Human-Computer Interaction
* Mental Health Monitoring
* Smart Surveillance Systems
* Customer Feedback Analysis
* Educational Systems

### 2) Data Collection

* Dataset Source - https://www.kaggle.com/datasets/msambare/fer2013?utm_source=chatgpt.com
The dataset consists of grayscale facial images categorized into seven emotion classes.
* Dataset Information - 
* Total Images: ~35,000+
* Image Size: 48 × 48 pixels
* Emotion Classes: 7
* Training Images: ~28,000+
* Testing Images: ~7,000+
